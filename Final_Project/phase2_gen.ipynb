```python
import warnings
import numpy as np
import os
import random
import re
from PIL import Image
import pandas as pd
import torch
from torch.utils.data import Dataset

import wandb
import evaluate
from transformers import (
    AutoProcessor,
    AutoModelForImageTextToText,
    Seq2SeqTrainer,
    Seq2SeqTrainingArguments,
    default_data_collator,
    GenerationConfig,
    EarlyStoppingCallback
)
from peft import LoraConfig, get_peft_model, TaskType
from pycocoevalcap.cider.cider import Cider

warnings.filterwarnings(
    "ignore",
    message="You are passing both `text` and `images` to `PaliGemmaProcessor`. The processor expects special image tokens in the text, as many tokens as there are images per each text. It is recommended to add `<image>` tokens in the very beginning of your text. For this call, we will infer how many images each text has and add special tokens."
)

# -------------------------------------------
# 0) WANDB oturumu & CSV yükleme + split
# -------------------------------------------
wandb.init(project="DI725_FinalProject", name="lora_r8__bestfilter")

# -------------------------------------------
# 1) Processor & Base Model
# -------------------------------------------
model_name = "google/paligemma-3b-pt-224"
processor  = AutoProcessor.from_pretrained(model_name, use_auth_token=True)
base_model = AutoModelForImageTextToText.from_pretrained(
    model_name, device_map="auto", use_auth_token=True
)
base_model.gradient_checkpointing_enable()
base_model.to("cuda")
base_model.eval()

# -------------------------------------------
# 2) Filtre: 3 kez üst üste aynı kelime
# -------------------------------------------
pattern_repeat = re.compile(r'(\b\w+\b)( \1){2,}')

def is_good_row(row):
    for i in range(1, 6):
        if pattern_repeat.search(row[f"caption_{i}"]):
            return False
    return True

train_df = train_df[train_df.apply(is_good_row, axis=1)].reset_index(drop=True)
val_df   = val_df[  val_df.apply(is_good_row, axis=1)].reset_index(drop=True)

# -------------------------------------------
# 3) Dataset tanımı
# -------------------------------------------
class RiscDynamicCaptionDataset(Dataset):
    def __init__(self, df, img_dir, processor, max_len=64):
        self.df      = df
        self.img_dir = img_dir
        self.proc    = processor
        self.max_len = max_len

    def __len__(self):
        return len(self.df)

    def __getitem__(self, idx):
        row = self.df.iloc[idx]
        img = Image.open(os.path.join(self.img_dir, row["image"])).convert("RGB")
        caps = [row[f"caption_{i}"] for i in range(1, 6)]
        cap  = random.choice(caps)
        enc = self.proc(
            images=img,
            text=cap,
            padding="max_length",
            truncation=True,
            max_length=self.max_len,
            return_tensors="pt"
        )
        return {
            "pixel_values":   enc.pixel_values.squeeze(0),
            "input_ids":      enc.input_ids.squeeze(0),
            "attention_mask": enc.attention_mask.squeeze(0),
            "labels":         enc.input_ids.squeeze(0).clone()
        }

img_dir  = "/content/resized"
train_ds = RiscDynamicCaptionDataset(train_df, img_dir, processor)
val_ds   = RiscDynamicCaptionDataset(val_df,   img_dir, processor)

# -------------------------------------------
# 4) LoRA Konfigürasyonu & PEFT Model
# -------------------------------------------
lora_cfg = LoraConfig(
    task_type=TaskType.SEQ_2_SEQ_LM,
    r=8,
    lora_alpha=16,
    lora_dropout=0.1,
    target_modules=["q_proj", "v_proj"]
)
model_lora = get_peft_model(base_model, lora_cfg)
model_lora.print_trainable_parameters()

# -------------------------------------------
# 5) GenerationConfig oluştur
# -------------------------------------------
gen_config = GenerationConfig(
    max_new_tokens=64,          # çıktı uzunluğu sınırı
    num_beams=3,                # beam search genişliği
    no_repeat_ngram_size=3,     # 2-gram tekrarını engelle
    early_stopping=True
)

# -------------------------------------------
# 6) TrainingArguments (step bazlı eval & save)
# -------------------------------------------
training_args = Seq2SeqTrainingArguments(
    output_dir="lora_r8_bestfilter",
    per_device_train_batch_size=2,
    gradient_accumulation_steps=4,
    learning_rate=5e-5,
    lr_scheduler_type="linear",
    num_train_epochs=1,            # tek epoch
    max_steps= 4000,

    logging_steps=50,

    eval_strategy="steps",         # adım bazında değerlendirme
    eval_steps=4000,                # her 500 adımda
    save_strategy="steps",         # adım bazında checkpoint
    save_steps=4000,                # her 500 adımda
    load_best_model_at_end=True,   # en iyi checkpoint’e dön
    metric_for_best_model="eval_loss",
    greater_is_better=False,

    predict_with_generate=True,
    generation_config=gen_config,

    fp16=True,
    report_to="wandb"
)

# -------------------------------------------
# 7) Trainer tanımı
# -------------------------------------------
trainer = Seq2SeqTrainer(
    model=model_lora,
    args=training_args,
    train_dataset=train_ds,
    eval_dataset=val_ds,
    data_collator=default_data_collator,
    tokenizer=processor,
    callbacks=[EarlyStoppingCallback(early_stopping_patience=2, early_stopping_threshold=0.1)]
)

# -------------------------------------------
# 8) Eğitim
# -------------------------------------------
trainer.train()

# -------------------------------------------
# 9) Metrikleri hesapla & yazdır
# -------------------------------------------
# 1) Tahminleri al
# 1) Trainer.predict ile raw_preds al
raw_preds = trainer.predict(val_ds).predictions  # bu bir np.ndarray veya torch.Tensor

# 2) NumPy dizisine dönüştür
if isinstance(raw_preds, np.ndarray):
    arr = raw_preds
else:
    arr = raw_preds.cpu().numpy()

# 3) Python listesine çevir ve her token id’i clamp et
tokenizer = processor.tokenizer
vocab_size = tokenizer.vocab_size
pad_id     = tokenizer.pad_token_id or tokenizer.eos_token_id

def clamp_tok(t):
    t = int(t)
    if t < 0 or t >= vocab_size:
        return pad_id
    return t

preds_list = [
    [clamp_tok(tok) for tok in seq.tolist()]  # seq zaten 1D array/tensor
    for seq in arr
]

# 4) Şimdi decode et
decoded_preds = tokenizer.batch_decode(
    preds_list,
    skip_special_tokens=True
)

refs = val_df[[f"caption_{i}" for i in range(1,6)]].values.tolist()

bleu_res   = evaluate.load("bleu"  ).compute(predictions=decoded_preds, references=refs)["bleu"]
meteor_res = evaluate.load("meteor").compute(predictions=decoded_preds, references=refs)["meteor"]
rouge_res  = evaluate.load("rouge" ).compute(predictions=decoded_preds, references=refs)["rougeL"]

cider_scorer = Cider()
cider_score, _ = cider_scorer.compute_score(
    {i: refs[i]           for i in range(len(refs))},
    {i: [decoded_preds[i]] for i in range(len(decoded_preds))}
)

print(f"BLEU:    {bleu_res:.4f}")
print(f"METEOR:  {meteor_res:.4f}")
print(f"ROUGE-L: {rouge_res:.4f}")
print(f"CIDEr:   {cider_score:.4f}")

# -------------------------------------------
# 10) CSV'e kaydet: pred & gt
# -------------------------------------------
out_rows = []
for i, img_name in enumerate(val_df["image"]):
    row = {"image": img_name, "pred": decoded_preds[i]}
    for j, gt in enumerate(refs[i], start=1):
        row[f"gt_{j}"] = gt
    out_rows.append(row)

pd.DataFrame(out_rows).to_csv("lora_r8_bestfilter_results.csv", index=False)
print("Saved predictions + GT to lora_r8_bestfilter_results.csv")
```
